```python
import numpy as np

# Roll the die 1000 times
rolls = np.random.choice([1, 2, 3, 4, 5, 6], size=1000)

# Condition: number > 3
condition = rolls > 3
condition_count = np.sum(condition)

# Event: even number AND number > 3
event = (rolls % 2 == 0) & condition
event_count = np.sum(event)

# Experimental conditional probability P(Even | >3)
if condition_count > 0:
    probability = event_count / condition_count
else:
    probability = 0

print("Total rolls:", len(rolls))
print("Condition satisfied (number > 3):", condition_count)
print("Event satisfied (even and > 3):", event_count)
print("Experimental Conditional Probability (Even | >3):", probability)
```
